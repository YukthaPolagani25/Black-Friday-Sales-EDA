#### Observation

**The `product_category_2` and `product_category_3` columns contain product category codes rather than continuous numerical measurements. The observed values range from 2 to 18 in `product_category_2` and from 3 to 18 in `product_category_3`.**

**Since these columns represent product categories, mean or median imputation is not appropriate because it could introduce an artificial category value. The missing values will therefore be treated as a separate category indicating that no additional product category was recorded.**

```python
df['product_category_2'] = df['product_category_2'].fillna(0)         # treating or handling the missing values
df['product_category_3'] = df['product_category_3'].fillna(0)
```

```python
df.isnull().sum()        
```

#### Observation

**The missing values in `product_category_2` and `product_category_3` were replaced with 0, which represents that no additional product category was recorded for that transaction. After treatment, all 12 columns contain zero missing values. No rows were removed because of missing values, so the dataset retains all 550,068 records.**

```python
df.duplicated().sum()      # Check duplicates
```

#### Observation

**The dataset contains 0 duplicate records. Therefore, no duplicate rows were removed, and all 550,068 records are retained for further analysis.**

```python
Q1 = df['purchase'].quantile(0.25)            # Outlier Detection
Q3 = df['purchase'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

#### Observation: Output Interpretation
**- Q1 = 5,823 is 25% of purchases are at or below ₹5,823.**

**- Q3 = 12,054 is 75% of purchases are at or below ₹12,054.**

**- IQR = 6,231 is middle 50% of purchase values lie between ₹5,823 and ₹12,054.**

**- Lower Bound = -3,523.5 → since purchase amounts cannot realistically be negative, there are no lower-side outliers.**
    
**- Upper Bound = 21,400.5 → purchase values above ₹21,400.5 will be treated as potential outliers.**

```python
outliers = df[(df['purchase']<lower_bound) | (df['purchase']>upper_bound)]        # Count the outliers
print("Number of Outliers:",len(outliers))
print("Percentage of outliers:", (len(outliers) / len(df)) * 100) 
```

```python
outliers['purchase'].describe()
```

```python
df = df[(df['purchase'] >= lower_bound) & (df['purchase'] <= upper_bound)]
```

```python
print("Rows after outlier treatment:", len(df))
print("Maximum purchase after treatment:", df['purchase'].max())
```

#### Observation

**The dataset initially contained 550,068 records. Using the IQR method, 2,677 potential outlier records were identified, representing approximately 0.49% of the dataset.The identified outliers were removed because they were above the calculated upper bound of ₹21,400.5. After treatment, the dataset contains 547,391 records, and the maximum purchase value is ₹21,399.
Only a small percentage of records were removed, so the majority of the original dataset is retained for further analysis.**

**- Hence, no class imbalance treatment is required for the target variable.**

### 5.3 Multivariate Analysis

Multivariate analysis is used to study relationships among three or more variables simultaneously. It helps us identify patterns and interactions that may not be visible when analyzing variables individually or in pairs.

```python
### 5.3.1 Age + Gender + Purchase

plt.figure(figsize=(10,5))

sns.barplot(
    data=df,
    x='age',
    y='purchase',
    hue='gender',
    estimator='mean'
)

plt.title("Average Purchase Amount by Age Group and Gender")
plt.xlabel("Age Group")
plt.ylabel("Average Purchase Amount")

plt.show()
```

#### Observation

**What we observed:**

**The grouped bar chart shows that the average purchase amount differs across age groups and between genders. For every age group shown, the average purchase amount for male customers is higher than that of female customers. The difference between male and female average purchase amounts is visible across all age groups.**

**Why this graph was used:**

**A grouped bar chart was used to analyze three variables together: age group, gender, and purchase amount. It helps compare the average purchase amount between genders across different age groups.**

```python
### 5.3.2 City + Gender + Purchase

plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x='city_category',
    y='purchase',
    hue='gender',
    estimator='mean'
)

plt.title("Average Purchase Amount by City Category and Gender")
plt.xlabel("City Category")
plt.ylabel("Average Purchase Amount")

plt.show()
```

**Why this graph was used:**
